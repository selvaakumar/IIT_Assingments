Problem 1  
Consider the following one-dimensional data points:
 X = {2,4,5,10,12,13}

Let K = 2, with initial centroids: C1 = 4 and C2 = 12

Perform one complete iteration of K-Means:
Calculate the distance of every point from both centroids. 
Assign each point to the nearest centroid. 
Calculate the new centroids. 
State the cluster assignments after the first iteration. 


```python
import numpy as np

x = np.array([2, 4, 5, 10, 12, 13])

centroids = np.array([4.0, 12.0])

distances = np.abs(x[:, np.newaxis] - centroids)

print("Distances:\n", distances)

labels = np.argmin(distances, axis=1)
print("Labels:\n", labels)

new_centroids = np.array([x[labels == i].mean() for i in range(len(centroids))])
print("New Centroids:\n", new_centroids)

print("--- Distances (rows: points, cols: [C1, C2]) ---")
for x, d, label in zip(x, distances, labels):
    print(
        f"x = {x:2d} | dist(C1) = {d[0]:.1f}, dist(C2) = {d[1]:.1f} | Assigned to Cluster {label + 1}"
    )

```

```python
import numpy as np

# 1. Data Setup
X = np.array([1, 2, 3, 8, 9, 10], dtype=float)
centroids = np.array([2.0, 9.0])

# 2. Assignment
distances = np.abs(X[:, np.newaxis] - centroids)
labels = np.argmin(distances, axis=1)

# 3. New Centroids
new_centroids = np.array([X[labels == k].mean() for k in range(len(centroids))])

# 4. SSE Calculation
sse_per_cluster = [
    np.sum((X[labels == k] - new_centroids[k]) ** 2)
    for k in range(len(new_centroids))
]
total_sse = sum(sse_per_cluster)

print(f"Cluster 1: {X[labels == 0].astype(int).tolist()}")
print(f"Cluster 2: {X[labels == 1].astype(int).tolist()}")
print(f"New Centroids: C1 = {new_centroids[0]:.1f}, C2 = {new_centroids[1]:.1f}")
print(f"SSE Cluster 1: {sse_per_cluster[0]}")
print(f"SSE Cluster 2: {sse_per_cluster[1]}")
print(f"Total SSE: {total_sse}")
```

Problem 4
Consider the following two-dimensional points in a Cartesian coordinate system:
A=(1,1)
B=(2,2)
C=(3,3)
D=(8,8)
E=(9,9)
F=(10,10)
Using K=2 clusters, initialize the algorithm with the following starting centroids:
Initial Centroid 1: C 
1
​	
 =(2,2)
Initial Centroid 2: C 
2
​	
 =(9,9)
Answer the following:
a) Perform one complete iteration of the K-Means algorithm by computing the Euclidean distance from every point to both initial centroids and assigning each point to its nearest cluster.
b) Calculate the updated coordinates for the new centroids based on these cluster assignments.
c) Calculate the total within-cluster Sum of Squared Errors (SSE) for the resulting clusters using the updated centroids.
d) Explain whether running a second iteration will alter the cluster assignments or centroid positions.

```python
labels_list = ["A", "B", "C", "D", "E", "F"]
X = np.array(
    [
        [1.0, 1.0],  
        [2.0, 2.0],  
        [3.0, 3.0],  
        [8.0, 8.0],  
        [9.0, 9.0],  
        [10.0, 10.0],  
    ]
)
initial_centroids = np.array([[2.0, 2.0], [9.0, 9.0]])
sq_distances = np.sum((X[:, np.newaxis, :] - initial_centroids) ** 2, axis=2)
print("Squared Distances:\n", sq_distances)
cluster_assignments = np.argmin(sq_distances, axis=1)
print("Assigned Clusters:\n", [labels_list[i] for i in cluster_assignments])


```

```python
print("--- Part (a): Assignments ---")
for name, pt, d, assigned in zip(
    labels_list, X, sq_distances, cluster_assignments
):
    print(
        f"Point {name} {pt.astype(int).tolist()} | "
        f"d^2(C1) = {d[0]:3.0f}, d^2(C2) = {d[1]:3.0f} | "
        f"Assigned to Cluster {assigned + 1}"
    )

for k in range(len(initial_centroids)):
    pts_in_k = [
        labels_list[i]
        for i, assigned in enumerate(cluster_assignments)
        if assigned == k
    ]
    print(f"Cluster {k + 1}: {pts_in_k}")


# Part (b): Calculate Updated Centroids

updated_centroids = np.array(
    [
        X[cluster_assignments == k].mean(axis=0)
        for k in range(len(initial_centroids))
    ]
)
print("\n--- Part (b): Updated Centroids ---")
for k, c in enumerate(updated_centroids):
    print(f"C{k + 1}_new: ({c[0]:.2f}, {c[1]:.2f})")
```

```python
print("\n--- Part (b): Updated Centroids ---")
for k, centroid in enumerate(updated_centroids):
    print(f"New C{k + 1}: ({centroid[0]:.1f}, {centroid[1]:.1f})")

# ----------------------------------------------------
# Part (c): Total Within-Cluster SSE
# ----------------------------------------------------
sse_per_cluster = []
for k in range(len(updated_centroids)):
    pts = X[cluster_assignments == k]
    sse_k = np.sum((pts - updated_centroids[k]) ** 2)
    sse_per_cluster.append(sse_k)

total_sse = sum(sse_per_cluster)

print("\n--- Part (c): Within-Cluster SSE ---")
for k, sse_val in enumerate(sse_per_cluster):
    print(f"SSE Cluster {k + 1}: {sse_val:.1f}")
print(f"Total SSE: {total_sse:.1f}")
```

```python
# ----------------------------------------------------
# Part (d): Convergence Check
# ----------------------------------------------------
has_converged = np.allclose(initial_centroids, updated_centroids)

print("\n--- Part (d): Convergence Check ---")
print(f"Centroids shifted: {not has_converged}")
if has_converged:
    print(
        "Result: Centroids are identical to initial centroids. Next iteration will NOT change assignments."
    )
else:
    print("Result: Centroids shifted; subsequent iterations may reassign points.")
```

Problem 5:

1. Which value of K would you select using the Elbow Method?
Select K=3 (or K=4).

Look at the marginal reduction in SSE (ΔSSE) for each step:
From K=1→K=2: reduction of 1200−650=550
From K=2→K=3: reduction of 650−420=230
From K=3→K=4: reduction of 420−350=70
From K=4→K=5: reduction of 350−330=20
From K=5→K=6: reduction of 330−320=10
The sharp drop flattens significantly after K=3 (where the slope drops from 230 to 70), making K=3 the most prominent "elbow" point. A case can also be made for K=4 where the subsequent gains drop to negligible values (≤20).

2. Why would selecting K=6 simply because it has the lowest SSE be misleading?
Monotonic Decrease of SSE: As K increases, SSE will always decrease (or remain constant) because adding more cluster centers places centroids closer to their assigned points by definition.
Trivial / Overfitting Solution: If this logic were followed to its extreme, setting K=N (where every data point is its own cluster centroid) would yield SSE=0. This provides no meaningful data aggregation or summarization, resulting in complete overfitting.
Diminishing Returns: Going from K=5 to K=6 only reduces SSE by 10 (less than 1% of the original variance), which adds model complexity without capturing genuine underlying cluster structure.

3. What does the sharp reduction in SSE from K=1 to K=3 indicate?
True Underlying Structure: It indicates that the dataset contains distinct, well-separated natural groupings rather than uniformly distributed noise.
Capturing Major Variance: Moving from a single global mean (K=1) to 3 centroids (K=3) captures the primary geometric separation of the data, resolving the bulk of the dispersion (SSE drops by nearly 65%, from 1200 down to 420).

Problem 6:

Mathematical Certainty: As the number of clusters (K) increases, the Sum of Squared Errors (SSE) will mathematically always decrease (or remain constant). Adding more centroids inherently forces data points to be closer to their assigned center.

Invalid Comparison: You cannot directly compare the raw SSE of two runs with different K values to determine which model is "better." Run B's lower SSE (180 vs 250) is simply the expected geometric outcome of increasing K from 2 to 3, not proof of a superior clustering structure.

Proper Evaluation: To determine if K=3 is genuinely better than K=2, you must evaluate the rate of decrease across a wider range of K values using the Elbow Method, or apply a metric that accounts for model complexity and cluster separation, such as the Silhouette Score.
