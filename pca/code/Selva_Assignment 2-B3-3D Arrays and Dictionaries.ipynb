```python
# Question 9  

Create a dictionary of five products.

Example

    Laptop
    Mouse
    Keyboard
    Monitor
    Printer

Each product should contain

    Product Name
    Price
    Quantity

Write a program to

    Display all products.
    Print the product with the highest price.
    Calculate the total inventory value

(Formula:  Inventory Value = Price × Quantity)
```

```python
products = {
    1: {
        "Product Name": "Laptop",
        "Price": 55000,
        "Quantity": 10
    },
    2: {
        "Product Name": "Mouse",
        "Price": 500,
        "Quantity": 50
    },
    3: {
        "Product Name": "Keyboard",
        "Price": 1200,
        "Quantity": 30
    },
    4: {
        "Product Name": "Monitor",
        "Price": 15000,
        "Quantity": 15
    },
    5: {
        "Product Name": "Printer",
        "Price": 8000,
        "Quantity": 8
    }
}

```

```python
#Display all products.
print(products)
```

```python
#Print the product with the highest price.
def hig_price(prod):
    High_price = 0
    for i in prod:
        if products[i]['Price'] > High_price:
            High_price = products[i]['Price']
    return High_price
print(hig_price(products))
```

```python
#Calculate the total inventory value
def total_inv(prod):
    Total = 0
    for i in prod:
        Total = products[i]['Price'] * products[i]['Quantity']
    return Total
print(total_inv(products))
```

```python
# Question 10

Create a dictionary of five students and their marks.

Example

    Alice : 85
    Bob : 92
    Charlie : 78
    David : 88
    Emma : 95

Write a Python program to

    Find the student with the highest marks.
    Find the student with the lowest marks.
    Calculate the average marks.
    Display students who scored more than 85.
```

```python
mark = {
    "Alice" : 85,
    "Bob" : 92,
    "Charlie" : 78,
    "David" : 88,
    "Emma" : 95
}
```

```python
#Find the student with the highest marks.

def hig_mark(prod):
    High_mark = 0
    for i in prod.values():
        if i > High_mark:
            High_mark = i
    return High_mark
print(hig_mark(mark))
```

```python
#Find the student with the lowest marks.
def low_mark(prod):
    Low_mark = 100
    for i in prod.values():
        if i < Low_mark:
            Low_mark = i
    return Low_mark
print(low_mark(mark))
```

```python
#Calculate the average marks.
def avg_mark(prod):
    mark = 0
    for i in prod.values():
        mark = mark + i
    mark = mark/len(prod)
    return mark
print("Average Mark : ",avg_mark(mark))
```

```python
#Display students who scored more than 85.
for name,mark in mark.items():
    if mark > 85:
        print("Name of Student mark more than 85 :", name)
```
