Task 10 — Broadcasting
Create:
X = np.ones((3, 4))
and:
b = np.array([10, 20, 30, 40])
Calculate: X + b
Then try: X * b
Question: Why can a (4,) tensor be combined with a (3,4) tensor?

```python
X = np.ones((3, 4))
b = np.array([10, 20, 30, 40])

addition = X + b
Multiply = X * b
```

1Dimension Alignment: NumPy aligns the shapes starting from the right side (trailing dimensions):X Shape: (3, 4) 
b Shape: (   4)
The Compatibility Check: Two axes are compatible if they are equal, or if one of them is 1. In this case, the trailing dimensions match perfectly (4 and 4).

Virtual Stretching: Because b lacks a first dimension, NumPy virtually adds a dimension of size 1 to its front, treating it as (1, 4). It then replicates (or "broadcasts") that single row down 3 times until it perfectly matches the (3, 4) footprint of X.

Task 11 — Broadcasting with Sensor Data
Suppose:
X.shape = (100, 10)
represents:
100 samples × 10 sensors
The calibration offset for each sensor is:
[1.2, 0.8, 1.1, 0.5, 0.9, 1.0, 0.7, 1.3, 0.6, 0.4]
Create the tensor and subtract the calibration offsets.
Your output should remain:
(100, 10)
Goal: See broadcasting in an actual engineering problem.

```python
import numpy as np


X = np.random.uniform(low=10.0, high=50.0, size=(100, 10))


offset = np.array([1.2, 0.8, 1.1, 0.5, 0.9, 1.0, 0.7, 1.3, 0.6, 0.4])

calibrated_X = X - offset

print("Original Sensor Data Shape: ", X.shape)
print("Calibration Offset Shape:   ", offset.shape)
print("Calibrated Data Shape:       ", calibrated_X.shape)


print("\n--- Verification on Sensor 0 (First 3 samples) ---")
print("Original:  ", X[:3, 0])
print("Calibrated:", calibrated_X[:3, 0])

```

This works flawlessly because of your trailing dimension alignment rule:
X Shape: (100, 10)
offset Shape: (     10)
Because the rightmost dimensions match (10 and 10), NumPy virtually duplicates the 10-element offset array 100 times downwards.This means Sensor 0's value (1.2) is subtracted from all 100 entries in Column 0, Sensor 1's value (0.8) is subtracted from all 100 entries in Column 1, and so on. You correct 1,000 data points in a single line of code without a single slow for loop.

Task 13 — Normalize a Tensor
Create sensor data:
X = np.random.randn(100, 20, 5)
Calculate the mean and standard deviation for each sensor.
Normalize the tensor using:
 
Then verify:
mean ≈ 0
standard deviation ≈ 1

```python
mean = X.mean(axis=(0, 1), keepdims=True)
std = X.std(axis=(0, 1), keepdims=True)

X_normalized = (X - mean) / std


post_mean = X_normalized.mean(axis=(0, 1))
post_std = X_normalized.std(axis=(0, 1))


print(f"Original shape:            {X.shape}")
print(f"Calculated mean shape:     {mean.shape} (due to keepdims=True)")
print(f"Normalized shape:          {X_normalized.shape}\n")

print("--- Verification per Sensor (Should be Mean ≈ 0, Std ≈ 1) ---")
for i in range(5):
    print(f"Sensor {i} -> Mean: {post_mean[i]:.6f} | Std Dev: {post_std[i]:.6f}")
```

Task 15 — Concatenation
Suppose a model produces two feature tensors:
Visual features:
(32, 128)

Sensor features:
(32, 64)
Combine them into a single feature tensor.
Expected result:
(32, 192)
Use tensor concatenation.
Then explain:
Why can we concatenate along the feature dimension but not arbitrarily along every dimension?

```python
import numpy as np

visual_features = np.random.randn(32, 128)
sensor_features = np.random.randn(32, 64)


combined_features = np.concatenate((visual_features, sensor_features), axis=1)


print(f"Visual features shape: {visual_features.shape}")
print(f"Sensor features shape: {sensor_features.shape}")
print(f"Combined shape:        {combined_features.shape}")  
```

1. The "All Other Dimensions Must Match" Rule: When you concatenate along axis=1 (the features), NumPy stitches the arrays side-by-side. This works because both tensors agree on axis=0 (they both have exactly 32 samples). Every single sample row matches up perfectly.
2. Why axis=0 Fails Here: If you attempted to concatenate them along axis=0 (stacking them vertically), NumPy would try to place a 64-column matrix directly beneath a 128-column matrix. This creates a structural mismatch—tensors must be uniform blocks (matrices cannot have jagged rows of varying widths).

Task 16 — Practical Mini Challenge: Sensor Dataset
Create a synthetic dataset representing:
500 machines
60 time steps
8 sensors

Therefore:

X.shape = (500, 60, 8)

Perform the following:
Extract machine 10. 
Extract the first 20 time steps. 
Extract sensor 5 from all machines. 
Calculate the average value of every sensor. 
Calculate the average value at every time step. 
Find the machine with the highest average sensor reading. 
Normalize each sensor. 
Split the tensor into: 
Training: 400 machines 
Testing: 100 machines 


```python
import numpy as np

np.random.seed(42)

X = np.random.randn(500, 60, 8)

machine_10 = X[10, :, :]
first_20_steps = X[:, :20, :]
sensor_5_all = X[:, :, 5]

avg_per_sensor = X.mean(axis=(0, 1))
avg_per_timestep = X.mean(axis=(0, 2))

machine_means = X.mean(axis=(1, 2))
highest_avg_machine_idx = np.argmax(machine_means)

sensor_means = X.mean(axis=(0, 1), keepdims=True)
sensor_stds = X.std(axis=(0, 1), keepdims=True)
X_normalized = (X - sensor_means) / sensor_stds

X_train = X_normalized[:400, :, :]
X_test = X_normalized[400:, :, :]

print(f"0. Initial dataset shape:                     {X.shape}")
print(f"1. Machine 10 shape:                          {machine_10.shape}")
print(f"2. First 20 time steps shape:                 {first_20_steps.shape}")
print(f"3. Sensor 5 across all machines shape:        {sensor_5_all.shape}")
print(f"4. Average per sensor shape:                  {avg_per_sensor.shape}")
print(f"5. Average per time step shape:               {avg_per_timestep.shape}")
print(f"6. Machine ID with highest average reading:   #{highest_avg_machine_idx}")
print(f"7. Normalized dataset shape:                  {X_normalized.shape}")
print(f"8. Train split shape | Test split shape:      {X_train.shape} | {X_test.shape}")
```
