Part A — Data Understanding
1. Load the dataset.

```python
import pandas as pd 
import numpy as np 

df = pd.read_csv('../data/ecommerce_orders_raw.csv')
df.head()
```

```python
#3. Identify the data types of all columns.
df.info()
```

```python
#4. Identify missing values.
df.isnull().sum()
```

```python
#5. Identify duplicate orders.
df.duplicated().sum()
```

6. Examine the unique values in:
o City
o Product Category
o Payment Method

```python
print(df['City'].unique())
df['City'] = df['City'].str.strip().str.title()
print(df['City'].unique())

#product category column
print(df['Product_Category'].unique())

#payment method column
print(df['Payment_Method'].unique())
```

Part B — Data Cleaning
7. Handle missing ratings.

```python
print("No of missing values in Rating column:", df['Rating'].isna().sum())
print("Unique values in Rating column:", df['Rating'].unique())

df['Rating'].fillna(df['Rating'].median(), inplace=True)
df['Rating'] = df['Rating'].astype(int)
df['Rating'].unique()
```

8. Standardize city names.

```python
df['City'].unique()
```

```python
#9. Identify invalid quantities.
print(df['Quantity'].unique())
df['Quantity'] = df['Quantity'].astype(int)
print(df['Quantity'].unique())
df[df['Quantity'] < 0]
#Quntity will not be in negative value so it will be positive value. So we will replace the negative values with positive values using abs() function.
df['Quantity'] = df['Quantity'].abs()
print(df['Quantity'].unique())

```

```python
#10. Handle duplicate orders.

print("Number of duplicate orders removed:", df.duplicated().sum())
df.drop_duplicates(inplace=True)
print("Number of duplicate orders after removal:", df.duplicated().sum())

```

```python
#11. Convert Order_Date into an appropriate datetime format.

df['Order_Date'] = df['Order_Date'].astype('datetime64[ns]')
df['Order_Date'] = df['Order_Date'].dt.strftime('%Y-%m-%d')
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Order_Date'].head()
```

Part C — Feature Engineering
12. Create:
Gross_Amount = Quantity × Unit_Price

```python
df['Gross_Amount'] = df['Quantity'].astype(int) * df['Unit_Price'].astype(float)
df['Gross_Amount'].head()
```

```python
df['Unit_Price']
```

13. Create:
Discount_Amount = Gross_Amount × Discount / 100

```python
df['Discount_Amount'] = (df['Gross_Amount'] * (df['Discount'].astype(float) / 100)).round(2)
df['Discount_Amount'].head()
```

14. Create:
Net_Amount = Gross_Amount - Discount_Amount


```python
df['Net_Amount'] = df['Gross_Amount'] - df['Discount_Amount']
df['Net_Amount'] = df['Net_Amount'].round(2)
df['Net_Amount'].head()
```

15. Create a Rating_Category column:
1–2 → Poor
3 → Average
4 → Good
5 → Excellent

```python
df['Rating'].head()
mapping = {1: 'Poor', 2: 'Fair', 3: 'Good', 4: 'Very Good', 5: 'Excellent'}
df['Rating_Category'] = df['Rating'].map(mapping)
df['Rating_Category'].head()
```

16. Extract the following from Order_Date:
• Year
• Month
• Day
• Day of Week

```python
df['Order_Date'].head()
df['Year'] = df['Order_Date'].dt.year
df['Month'] = df['Order_Date'].dt.month
df['Day'] = df['Order_Date'].dt.day
df['Dayof_Week'] = df['Order_Date'].dt.day_name()
df[['Order_Date', 'Year', 'Month', 'Day', 'Dayof_Week']].head()
```

Part D — Business Analysis

```python
#17. Calculate total revenue.
total_revenue = df['Net_Amount'].sum().round(2)
print("Total Revenue: ", total_revenue)
```

```python
#18. Calculate total number of orders.
total_orders = df['Order_ID'].nunique()
print("Total Number of Orders: ", total_orders)
```

```python
#19. Find the average order value.
average_order_value = total_revenue / total_orders
print("Average Order Value: ", average_order_value.round(2))
```

```python
#20. Find total revenue by product category.
total_revenue_by_category = df.groupby('Product_Category')['Net_Amount'].sum().round(2)
print("Total Revenue by Product Category:")
print(total_revenue_by_category)
```

```python
df.columns
```

```python
#21. Find the top 10 products by revenue.
top_10_products = df.groupby('Product')['Net_Amount'].sum().round(2).nlargest(10)
print("Top 10 Products by Revenue:")
print(top_10_products)
```

```python
#2. Find the city generating the highest revenue.
highest_revenue_city = df.groupby('City')['Net_Amount'].sum().round(2).nlargest(1)
print("City generating the highest revenue:")
print(highest_revenue_city)
```

```python
df['Payment_Method'].value_counts()
```

```python
#23. Find the most frequently used payment method.
most_frequent_payment_method = df['Payment_Method'].value_counts().idxmax()
print("Most Frequently Used Payment Method:")
print(most_frequent_payment_method)
```

```python
#24. Find the average rating for each product category.
average_rating_by_category = df.groupby('Product_Category')['Rating'].mean().round(2)
print("Average Rating by Product Category:")
print(average_rating_by_category)
```

```python
#25. Calculate monthly revenue.
monthly_revenue = df.groupby(df['Order_Date'].dt.to_period('M'))['Net_Amount'].sum().round(2)
print("Monthly Revenue:")
print(monthly_revenue)
```

```python
#26. Identify the highest-revenue month.
highest_revenue_month = monthly_revenue.idxmax()
print("Highest Revenue Month:", highest_revenue_month)
```

```python
#27. Find the top 10 customers by total spending.
top_10_customers = df.groupby('Customer_ID')['Net_Amount'].sum().round(2).nlargest(10)
print("Top 10 Customers by Total Spending:")
print(top_10_customers)
```

```python
df.columns
```

```python
#28. Compare revenue generated by different cities.
revenue_by_city = df.groupby('City')['Net_Amount'].sum().round(2)
print("Revenue by City:")
print(revenue_by_city)
```

```python
#29. Find which product category has the highest average order value.
average_order_value_by_category = df.groupby('Product_Category')['Net_Amount'].mean().round(2)
highest_avg_order_value_category = average_order_value_by_category.idxmax()
print("Product Category with Highest Average Order Value:")
print(highest_avg_order_value_category, "with an average order value of", average_order_value_by_category[highest_avg_order_value_category])
```
