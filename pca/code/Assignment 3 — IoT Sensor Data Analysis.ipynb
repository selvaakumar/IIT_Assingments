19. Create a Battery_Status column:
Battery Status
>= 50 Healthy
20–49 Moderate
<20 Critical

```python
df['Battery_Status'] = 'Healthy'

for i in range(len(df)):
    if df.loc[i, 'Battery_Level'] < 20:
        df.loc[i, 'Battery_Status'] = 'Critical'
    elif df.loc[i, 'Battery_Level'] < 50:
        df.loc[i, 'Battery_Status'] = 'Moderate'
    else:
        df.loc[i, 'Battery_Status'] = 'Healthy'

df['Battery_Status'].unique()
df[['Battery_Level', 'Battery_Status']].head(10)
```

```python
#20. Create a Temperature_Status column.
df['Temperature_Status'] = 'Normal'

for i in range(len(df)):
    if df.loc[i, 'Temperature'] > 75:
        df.loc[i, 'Temperature_Status'] = 'VeryHigh'
    elif df.loc[i, 'Temperature'] > 60:
        df.loc[i, 'Temperature_Status'] = 'High'
    elif df.loc[i, 'Temperature'] < 10:
        df.loc[i, 'Temperature_Status'] = 'Low'
    else:
        df.loc[i, 'Temperature_Status'] = 'Normal'

df['Temperature_Status'].unique()
df[['Temperature', 'Temperature_Status']].head(10)
```

```python
print(df['Vibration'].max())
print(df['Vibration'].min())
print(df['Vibration'].mean())
print(df['Vibration'].median())
```

```python
#21. Create a Vibration_Status column.
df['Vibration_Status'] = 'Normal'

for i in range(len(df)):
    if df.loc[i, 'Vibration'] > 10:
        df.loc[i, 'Vibration_Status'] = 'VeryHigh'
    elif df.loc[i, 'Vibration'] > 6:
        df.loc[i, 'Vibration_Status'] = 'High'
    elif df.loc[i, 'Vibration'] < 1:
        df.loc[i, 'Vibration_Status'] = 'Low'
    else:
        df.loc[i, 'Vibration_Status'] = 'Normal'

df['Vibration_Status'].unique()
df[['Vibration', 'Vibration_Status']].head(10)
```

22. Create an overall Machine_Health column:
Normal
Warning
Critical

```python
#Create an overall Machine_Health column:
df['Machine_Health'] = 'Normal'
for i in range(len(df)):
    if  df.loc[i, 'Temperature_Status'] == 'VeryHigh' or df.loc[i, 'Vibration_Status'] == 'VeryHigh':
        df.loc[i, 'Machine_Health'] = 'Critical'
    elif df.loc[i, 'Temperature_Status'] == 'High' or df.loc[i, 'Vibration_Status'] == 'High':
        df.loc[i, 'Machine_Health'] = 'Moderate'
    else:
        df.loc[i, 'Machine_Health'] = 'Normal'

df['Machine_Health'].unique()
df[['Temperature_Status', 'Vibration_Status', 'Machine_Health']].head(10)
```

```python
#23. Find devices that have experienced critical conditions more than five times.
critical_devices = df[df['Machine_Health'] == 'Critical']
critical_devices.groupby('Device_ID').size()[critical_devices.groupby('Device_ID').size() > 5]
```

```python
#24. Find the factory with the highest number of abnormal sensor readings.
abnormal_readings = df[(df['Temperature_Status'] == 'VeryHigh') | (df['Vibration_Status'] == 'VeryHigh')]
abnormal_readings.groupby('Device_ID').size().idxmax()
```
