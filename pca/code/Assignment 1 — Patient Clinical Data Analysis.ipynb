12. Correct or handle patients whose age is outside a reasonable range.

```python
# 12. Correct or handle patients whose age is outside a reasonable range
age_col = 'Age'
if age_col not in df.columns:
    print(f"Column '{age_col}' not found in dataframe.")
else:
    # Identify invalid ages (negative or unrealistically high)
    invalid_mask = (df[age_col] < 0) | (df[age_col] > 120)
    total_invalid = invalid_mask.sum()
    print(f"Total invalid age entries: {total_invalid}")
    if total_invalid > 0:
        print('\nSample invalid age rows:')
        display(df.loc[invalid_mask].head(10))
        # Set invalid ages to NaN for safe imputation
        df.loc[invalid_mask, age_col] = np.nan
        # Impute with median age (rounded to int)
        median_age = int(df[age_col].median(skipna=True))
        df[age_col] = df[age_col].fillna(median_age).astype(int)
        print(f"\nReplaced invalid ages with median age: {median_age}")
        remaining_invalid = ((df[age_col] < 0) | (df[age_col] > 120)).sum()
        print(f"Invalid age entries after correction: {remaining_invalid}")
    else:
        print('No invalid ages found (within 0-120).')
```

Part C — Data Transformation
15. Create a new column called Age_Group:
< 18 Child
18–40 Young Adult
41–60 Middle Age
> 60 Senior

```python
import numpy as np
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['Age_Group'] = np.nan
for i in range(len(df)):
    age = df.loc[i, 'Age']
    if pd.isna(age):
        df.loc[i, 'Age_Group'] = np.nan
    elif age < 18:
        df.loc[i, 'Age_Group'] = 'Child'
    elif age <= 40:
        df.loc[i, 'Age_Group'] = 'Young Adult'
    elif age <= 60:
        df.loc[i, 'Age_Group'] = 'Middle Aged'
    else:
        df.loc[i, 'Age_Group'] = 'Senior'
```

17. Create a High_Risk column based on suitable clinical conditions.

```python

Age_Group = df['Age_Group'] == 'Senior'
Diagnosis = df['Diagnosis'].astype(str).str.strip().str.title().isin(['Cancer','Heart Disease'])
Emergency = df['Admission_Type'].astype(str).str.strip().str.lower() == 'emergency'

high_risk = Age_Group | Diagnosis | Emergency

df['High_Risk'] = high_risk.map({True: 'Yes', False: 'No'})
```

18. Create a Cost_Per_Day column using treatment cost and length of stay.

```python
cost = df['Treatment_Cost'].astype(int)
days = df['Length_of_Stay'].astype(int)
df['Cost_Per_Day'] = cost / days
df['Cost_Per_Day'] = df['Cost_Per_Day'].round(2)
df.head()
```
