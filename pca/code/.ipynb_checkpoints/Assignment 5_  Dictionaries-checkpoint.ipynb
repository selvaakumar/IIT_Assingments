```python
Q30. Generate an employee report
```

```python
total_employees = len(employees)

missing_email = 0
missing_phone = 0
total_salary = 0
highest_salary = 0
lowest_salary = float("inf")

for details in employees.values():

    if details["Email"] is None:
        missing_email += 1

    if details["Phone"] is None:
        missing_phone += 1

    salary = details["Salary"]
    total_salary += salary

    if salary > highest_salary:
        highest_salary = salary

    if salary < lowest_salary:
        lowest_salary = salary

average_salary = total_salary / total_employees

print("Employee Report")
print("----------------")
print("Total Employees:", total_employees)
print("Missing Email:", missing_email)
print("Missing Phone:", missing_phone)
print("Average Salary:", average_salary)
print("Highest Salary:", highest_salary)
print("Lowest Salary:", lowest_salary)
```

```python
Q31. Generate report department-wise
```

```python
department_employees = {}

for details in employees.values():

    department = details["Department"]

    if department not in department_employees:
        department_employees[department] = []

    department_employees[department].append(details["Name"])


for department, employees_list in department_employees.items():

    print("\n" + department)
    print("-" * len(department))

    for name in employees_list:
        print(name)
```

```python
Q32. Employees sorted by salary — highest to lowest
```

```python
Q36. Department with highest average salary
```

```python
department_salary = {}
department_count = {}
for details in employees.values():

    department = details["Department"]
    salary = details["Salary"]

    if department not in department_salary:
        department_salary[department] = 0
        department_count[department] = 0

    department_salary[department] += salary
    department_count[department] += 1

department_average = {}

for department in department_salary:

    department_average[department] = (
        department_salary[department]
        / department_count[department]
    )

highest_department = max(
    department_average,
    key=department_average.get
)

print("Department:", highest_department)
print("Average Salary:", department_average[highest_department])
```
