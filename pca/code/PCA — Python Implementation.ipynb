7. Given the covariance matrix:
C = np.array([
[4, 2],
[2, 3]
])
Using NumPy:
Calculate the eigenvalues.
• Calculate the eigenvectors.
• Identify the eigenvector corresponding to the largest eigenvalue.

```python
import numpy as np

# Define the covariance matrix
C = np.array([
    [4, 2],
    [2, 3]
])

# 1. & 2. Calculate the eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(C)

# 3. Identify the eigenvector corresponding to the largest eigenvalue
largest_index = np.argmax(eigenvalues)
largest_eigenvalue = eigenvalues[largest_index]
principal_eigenvector = eigenvectors[:, largest_index]

# Print the results neatly
print("--- Calculations ---")
print(f"Eigenvalues: {eigenvalues}")
print(f"Eigenvectors (as columns):\n{eigenvectors}\n")

print("--- Selection ---")
print(f"Largest Eigenvalue: {largest_eigenvalue:.4f}")
print(f"Corresponding Eigenvector: {principal_eigenvector}")
```

9. Given a covariance matrix:
C = np.array([
[6, 4],
[4, 3]
])
Using Python:
• Find the eigenvalues.
• Find the eigenvectors.
• Sort the eigenvalues from largest to smallest.
• Rearrange the eigenvectors accordingly.

```python
import numpy as np

# Define the covariance matrix
C = np.array([
[6, 4],
[4, 3]
])

# 1. & 2. Find the raw eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(C)

# 3. Sort the eigenvalues from largest to smallest
# np.argsort returns indices for ascending order; [::-1] reverses it to descending
sorted_indices = np.argsort(eigenvalues)[::-1]
sorted_eigenvalues = eigenvalues[sorted_indices]

# 4. Rearrange the eigenvectors accordingly
# Remember: Eigenvectors are columns, so we slice using the sorted indices along the columns axis
sorted_eigenvectors = eigenvectors[:, sorted_indices]

# Print the final arranged results neatly
print("--- Sorted Results (Largest to Smallest) ---")
print(f"Sorted Eigenvalues:\n{sorted_eigenvalues}\n")
print(f"Rearranged Eigenvectors (as columns):\n{sorted_eigenvectors}")
```

Use the following data:
X = np.array([
[2, 1],

[4, 3],
[5, 4],

])

```python
import numpy as np

# --- 1. Original Data ---
X = np.array([
[2, 1],
[3, 2],
[4, 3],
[5, 4],
[6, 5]
])

print("1. Original Data Shape:", X.shape)

# --- 2. Mean Centering ---
mean = np.mean(X, axis=0)
# Subtract the mean from the data points
X_centered = X - mean

print("\n2. Mean Centered Data:\n", X_centered)

# --- 3. Covariance Matrix ---
# Calculate sample covariance matrix (rowvar=False because features are in columns)
cov_matrix = np.cov(X_centered, rowvar=False)

print("\n3. Covariance Matrix:\n", cov_matrix)

# --- 4. Eigenvalues & Eigenvectors ---
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print("\n4. Raw Eigenvalues:", eigenvalues)
print("4. Raw Eigenvectors (Columns):\n", eigenvectors)

# --- 5. Sort Components ---
# Get indices that sort eigenvalues in descending order
sorted_indices = np.argsort(eigenvalues)[::-1]
sorted_eigenvalues = eigenvalues[sorted_indices]
sorted_eigenvectors = eigenvectors[:, sorted_indices]

print("\n5. Sorted Eigenvalues (Descending):", sorted_eigenvalues)
print("5. Sorted Eigenvectors:\n", sorted_eigenvectors)

# --- 6. Select Principal Components ---
n_components = 1
principal_components = sorted_eigenvectors[:, :n_components]

print(f"\n6. Selected Principal Component Matrix (k={n_components}):\n", principal_components)

# --- 7. Transform Data ---
# Project the centered data onto the selected principal components
X_transformed = np.dot(X_centered, principal_components)

print("\n7. Transformed Data (Reduced Dimension):\n", X_transformed)
```

11. Using the same dataset, project the centered data onto the first principal

Calculate the resulting one-dimensional representation.

```python
import numpy as np

# 1. Original Data

X = np.array([
[2, 1],
[3, 2],
[4, 3],
[5, 4],
[6, 5]
])
# 2. Mean Centering
X_centered = X - np.mean(X, axis=0)

# 3. Get Covariance and Vectors
cov_matrix = np.cov(X_centered, rowvar=False)
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

# 4. Extract the First Principal Component (PC1)
sorted_indices = np.argsort(eigenvalues)[::-1]
pc1_vector = eigenvectors[:, sorted_indices[0]]

# 5. Project onto PC1 (Reshaped vector to a column for matrix multiplication)
X_1D = np.dot(X_centered, pc1_vector.reshape(-1, 1))

print("First Principal Component Vector (PC1):")
print(pc1_vector)
print("\nResulting One-Dimensional Representation:")
print(X_1D)
```

12. Modify your PCA program so that the user can specify:
n_components = 1
and the program automatically:
• Selects the required eigenvectors.
• Projects the data.
• Returns the reduced data.

```python
import numpy as np

def run_pca(data_matrix, n_components=1):
 
    mean = np.mean(data_matrix, axis=0)
    data_centered = data_matrix - mean
    
    cov_matrix = np.cov(data_centered, rowvar=False)
    
    eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
    
    sorted_indices = np.argsort(eigenvalues)[::-1]
    sorted_eigenvectors = eigenvectors[:, sorted_indices]

    selected_components = sorted_eigenvectors[:, :n_components]
    
    reduced_data = np.dot(data_centered, selected_components)
    
    return reduced_data


X = np.array([
[2, 1],
[3, 2],
[4, 3],
[5, 4],
[6, 5]
])

n_components = 1


X_reduced = run_pca(X, n_components=n_components)

print(f"--- Reduced Data (n_components = {n_components}) ---")
print(X_reduced)
print(f"\nNew Output Data Shape: {X_reduced.shape}")
```

18. Create a dataset where one feature has a much larger numerical scale than the
other:
X = np.array([[1, 100],
[2, 200],
[3, 300],
[4, 400],
[5, 500]
])
Apply PCA:
• Without standardization.
• After standardization using StandardScaler.
Compare the results.

```python
X = np.array([[1, 100],
[2, 200],
[3, 300],
[4, 400],
[5, 500]
])
import numpy as np


def custom_pca(data, n_components=1):
    """A helper function to run the PCA projection pipeline."""
    mean = np.mean(data, axis=0)
    centered = data - mean
    cov_matrix = np.cov(centered, rowvar=False)
    eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
    
    # Sort descending
    idx = np.argsort(eigenvalues)[::-1]
    sorted_eigenvectors = eigenvectors[:, idx]
    
    selected_vecs = sorted_eigenvectors[:, :n_components]
    transformed = np.dot(centered, selected_vecs)
    
    return transformed, sorted_eigenvectors[:, 0], eigenvalues

# --- 2. PCA Without Standardization ---
transformed_raw, pc1_raw, ev_raw = custom_pca(X, n_components=1)

# --- 3. PCA With Standardization (StandardScaler math from scratch) ---
# Calculate column-wise means and sample standard deviations (ddof=1)
mean_X = np.mean(X, axis=0)
std_X = np.std(X, axis=0, ddof=1)

# Standardize: (x - mean) / std
X_standardized = (X - mean_X) / std_X

transformed_scaled, pc1_scaled, ev_scaled = custom_pca(X_standardized, n_components=1)

# --- 4. Print Results ---
np.set_printoptions(suppress=True, precision=4)
print("=== PCA WITHOUT STANDARDIZATION ===")
print("First Principal Component Vector (PC1 directional loading):")
print(pc1_raw)
print("Raw Eigenvalues (Variance per axis):", ev_raw)
print("1D Transformed Data Representation:\n", transformed_raw)

print("\n=== PCA WITH STANDARDIZATION ===")
print("First Principal Component Vector (PC1 directional loading):")
print(pc1_scaled)
print("Scaled Eigenvalues (Variance per axis):", ev_scaled)
print("1D Transformed Data Representation:\n", transformed_scaled)
```
