You are working as a business data analyst. The company stores customer, product, and order
information in separate datasets. Your task is to combine these datasets and perform a complete business
analysis. This assignment focuses heavily on merging and combining DataFrames.

```python
#1. Load all three CSV files.
import pandas as pd
import numpy as np
df_customer = pd.read_csv('../data/customers.csv')
df_product = pd.read_csv('../data/products.csv')
df_orders = pd.read_csv('../data/orders.csv')
```

```python
#2. Display the dimensions of each DataFrame.
print("Customer DataFrame dimensions:", df_customer.shape)
print("Product DataFrame dimensions:", df_product.shape)
print("Orders DataFrame dimensions:", df_orders.shape)
```

```python
#3. Identify the columns in each dataset.
print("Customer DataFrame columns:", df_customer.columns.tolist())
print("Product DataFrame columns:", df_product.columns.tolist())
print("Orders DataFrame columns:", df_orders.columns.tolist())
```

```python
#4. Identify the common columns between datasets.
common_columns = set(df_customer.columns) & set(df_orders.columns)
print("Common columns:", list(common_columns))
common_columns = set(df_product.columns) & set(df_orders.columns)
print("Common columns:", list(common_columns))
```

5. Determine how the three datasets are related.

customer contain customer details and product contains product details 
order will relate to both customer and product based on  customerID and ProductID

```python
#6. Check each DataFrame for missing values.
print("Missing values in Customer DataFrame:")
print(df_customer.isnull().sum())
print("Missing values in Product DataFrame:")
print(df_product.isnull().sum())
print("Missing values in Orders DataFrame:")
print(df_orders.isnull().sum())
```

```python
#7. Check for duplicate records.
print("Duplicate records in Customer DataFrame:", df_customer.duplicated().sum())
print("Duplicate records in Product DataFrame:", df_product.duplicated().sum())
print("Duplicate records in Orders DataFrame:", df_orders.duplicated().sum())
```

```python
#8. Check the validity of Customer_ID and Product_ID.
print("Valid Customer_IDs in Orders DataFrame:", df_orders['Customer_ID'].isin(df_customer['Customer_ID']).sum())
print("Valid Product_IDs in Orders DataFrame:", df_orders['Product_ID'].isin(df_product['Product_ID']).sum())
```

```python
#9. Convert Order_Date into datetime.
df_orders['Order_Date'] = pd.to_datetime(df_orders['Order_Date'])
```

```python
#0. Check whether every order corresponds to a valid customer.
print("Orders with valid Customer_IDs:", df_orders['Customer_ID'].isin(df_customer['Customer_ID']).sum())
print("Orders with invalid Customer_IDs:", df_orders['Customer_ID'].isin(df_customer['Customer_ID']).value_counts().get(False, 0))
```

```python
#11. Check whether every order corresponds to a valid product.
print("Orders with valid Product_IDs:", df_orders['Product_ID'].isin(df_product['Product_ID']).sum())
print("Orders with invalid Product_IDs:", df_orders['Product_ID'].isin(df_product['Product_ID']).value_counts().get(False, 0))
```

```python
#12. Merge orders with customers.
df_orders_customers = pd.merge(df_orders, df_customer, on='Customer_ID', how='left')
```

```python
#13. Merge the resulting DataFrame with products.
df_orders_customers_products = pd.merge(df_orders_customers, df_product, on='Product_ID', how='left')
```

```python
#14. Create a final consolidated DataFrame.
df_final = df_orders_customers_products.copy()
df_final.columns
```

15. Create:
Order_Value = Price × Quantity

```python
df_final['Order_Value'] = df_final['Quantity'] * df_final['Price']
df_final['Order_Value'] = df_final['Order_Value'].round(2)
df_final['Order_Value'].head()
```

```python
#16. Calculate total revenue.
total_revenue = df_final['Order_Value'].sum()
print("Total Revenue:", total_revenue)
```

```python
#17. Calculate average order value.
average_order_value = df_final['Order_Value'].mean()
print("Average Order Value:", average_order_value)
```

```python
#18. Find total revenue by product category.
total_revenue_by_category = df_final.groupby('Category')['Order_Value'].sum()
print("Total Revenue by Product Category:")
print(total_revenue_by_category)
```

```python
#19. Find total revenue by city.
total_revenue_by_city = df_final.groupby('City')['Order_Value'].sum()
print("Total Revenue by City:")
print(total_revenue_by_city)
```

```python
#20. Find the top 10 customers by spending.
top_customers = df_final.groupby('Customer_ID')['Order_Value'].sum().nlargest(10)
print("Top 10 Customers by Spending:")
print(top_customers)
```

```python
#21. Find the top 10 products by revenue.
top_products = df_final.groupby('Product_ID')['Order_Value'].sum().nlargest(10)
print("Top 10 Products by Revenue:")
print(top_products)
```

```python
#22. Find the most frequently purchased product.
most_frequent_product = df_final.groupby('Product_ID').size().idxmax()
print("Most Frequently Purchased Product:")
print(most_frequent_product)
```

```python
#23. Find the city with the highest number of customers.
city_with_most_customers = df_final.groupby('City').size().idxmax()
print("City with the Highest Number of Customers:")
print(city_with_most_customers)
```

```python
#24. Find the city generating the highest revenue.
city_with_highest_revenue = df_final.groupby('City')['Order_Value'].sum().idxmax()
print("City generating the highest revenue:")
print(city_with_highest_revenue)
```

25. Divide customers into age groups.
18–30
31–45
46–60
61+

```python
#Divide customers into age groups.
df_final['Age_Group'] = pd.cut(df_final['Age'], bins=[0, 18, 30, 45, 60, 100], labels=['Under 18', '18-30', '31-45', '46-60', 'Over 60'])
df_final['Age_Group'].value_counts()
```

```python
#26. Calculate revenue generated by each age group.
revenue_by_age_group = df_final.groupby('Age_Group')['Order_Value'].sum()
print("Revenue by Age Group:")
print(revenue_by_age_group)
```

```python
#27. Compare spending between male and female customers.
spending_by_gender = df_final.groupby('Gender')['Order_Value'].sum()
print("Spending by Gender:")
print(spending_by_gender)   
```

```python
#28. Identify high-value customers.
high_value_customers = df_final.groupby('Customer_ID')['Order_Value'].sum().nlargest(10)
print("High-Value Customers:")
print(high_value_customers)
```

```python
#29. Determine the percentage of customers who have placed more than one order.
customer_order_counts = df_final.groupby('Customer_ID').size()
percentage_multiple_orders = (customer_order_counts > 1).mean() * 100
print(f"Percentage of customers with multiple orders: {percentage_multiple_orders:.2f}%")
```

```python
#30. Create a customer spending ranking.
customer_spending_ranking = df_final.groupby('Customer_ID')['Order_Value'].sum().rank(ascending=False)
print("Customer Spending Ranking:")
print(customer_spending_ranking)
```

```python
#31. Identify the top 10% of customers based on total spending.
top_10_percent_customers = df_final.groupby('Customer_ID')['Order_Value'].sum().nlargest(int(len(df_final['Customer_ID'].unique()) * 0.1))
print("Top 10% of Customers by Spending:")
print(top_10_percent_customers)
```

```python
#2. Find the most profitable product category for each city.
most_profitable_categories = df_final.groupby(['City', 'Category'])['Order_Value'].sum().groupby('City').idxmax()
print("Most Profitable Product Categories by City:")
print(most_profitable_categories)
```

```python
#33. Find the most popular product in each category.
most_popular_products = df_final.groupby(['Category', 'Product_ID']).size().groupby('Category').idxmax()
print("Most Popular Products by Category:")
print(most_popular_products)
```

```python
#34. Calculate monthly revenue.
monthly_revenue = df_final.groupby(df_final['Order_Date'].dt.to_period('M'))['Order_Value'].sum()
print("Monthly Revenue:")
print(monthly_revenue)
```

```python
#5. Identify the month with the highest revenue.
highest_revenue_month = monthly_revenue.idxmax()
print("Month with Highest Revenue:")
print(highest_revenue_month)    
```
