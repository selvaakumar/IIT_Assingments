```python
# Q21. Find the project with the highest funding.
highest_project = None
highest_funding = -1

for university in university_system["universities"].values():
    for faculty in university["faculties"].values():
        for department in faculty["departments"].values():
            for lab_name, lab in department.get("labs", {}).items():
                for project_name, project in lab["research_projects"].items():
                    if project["funding"] > highest_funding:
                        highest_funding = project["funding"]
                        highest_project = (project_name, project["funding"], lab_name)

print("Project:", highest_project[0])
print("Funding:", highest_project[1])
print("Lab:", highest_project[2])
```

```python
# Q22. Find total funding across all universities.
total_funding = 0

for university in university_system["universities"].values():
    for faculty in university["faculties"].values():
        for department in faculty["departments"].values():
            for lab in department.get("labs", {}).values():
                for project in lab["research_projects"].values():
                    total_funding += project["funding"]

print(total_funding)
```

```python
# Q25. Count total number of researchers in GTU_India.
# This counts researcher entries across all projects, including repeated names if any.
count = 0

for faculty in university_system["universities"]["GTU_India"]["faculties"].values():
    for department in faculty["departments"].values():
        for lab in department.get("labs", {}).values():
            for project in lab["research_projects"].values():
                count += len(project["researchers"])

print(count)
```

```python
# Q26. Find which lab has the maximum funding.
lab_funding = {}

for university in university_system["universities"].values():
    for faculty in university["faculties"].values():
        for department in faculty["departments"].values():
            for lab_name, lab in department.get("labs", {}).items():
                total = sum(project["funding"] for project in lab["research_projects"].values())
                lab_funding[lab_name] = lab_funding.get(lab_name, 0) + total

max_lab = max(lab_funding, key=lab_funding.get)
print("Lab:", max_lab)
print("Total funding:", lab_funding[max_lab])
```

```python
# Q27. Print projects grouped by status (ongoing/completed).
grouped = {"ongoing": [], "completed": []}

for university in university_system["universities"].values():
    for faculty in university["faculties"].values():
        for department in faculty["departments"].values():
            for lab in department.get("labs", {}).values():
                for name, project in lab["research_projects"].items():
                    grouped[project["status"]].append(name)

print("Ongoing:", grouped["ongoing"])
print("Completed:", grouped["completed"])
```

```python
# Q30. Write a function that searches for a project name and prints its full details.
def search_project(project_name):
    for university_name, university in university_system["universities"].items():
        for faculty_name, faculty in university["faculties"].items():
            for department_name, department in faculty["departments"].items():
                for lab_name, lab in department.get("labs", {}).items():
                    if project_name in lab["research_projects"]:
                        print("University:", university_name)
                        print("Faculty:", faculty_name)
                        print("Department:", department_name)
                        print("Lab:", lab_name)
                        print("Project:", project_name)
                        print("Details:", lab["research_projects"][project_name])
                        return

    print("Project not found")

# Example:
search_project("DeepVision")
```
